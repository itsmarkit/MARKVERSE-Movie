```python
import pandas as pd

def calculate_sales_summary(prices_list):
    """
    User-defined function to calculate sales statistics
    Parameters: prices_list - list of item prices
    Returns: dictionary with sales summary
    """
    if not prices_list:  # Check if list is empty
        return {
            'total_items': 0,
            'total_sales': 0.0,
            'average_sales': 0.0,
            'highest_sale': 0.0,
            'lowest_sale': 0.0
        }
    
    total_sales = sum(prices_list)
    average_sales = total_sales / len(prices_list)
    
    return {
        'total_items': len(prices_list),
        'total_sales': total_sales,
        'average_sales': average_sales,
        'highest_sale': max(prices_list),
        'lowest_sale': min(prices_list)
    }

def display_sales_table(sales_data):
    """
    Function to display sales summary in a formatted table using pandas
    Parameters: sales_data - dictionary with sales information
    """
    # Create DataFrame for better table formatting
    df = pd.DataFrame([
        ['Total Items Sold', sales_data['total_items']],
        ['Total Sales ($)', f"{sales_data['total_sales']:.2f}"],
        ['Average Sale ($)', f"{sales_data['average_sales']:.2f}"],
        ['Highest Sale ($)', f"{sales_data['highest_sale']:.2f}"],
        ['Lowest Sale ($)', f"{sales_data['lowest_sale']:.2f}"]
    ], columns=['Description', 'Value'])
    
    print("\n" + "="*50)
    print("           DAILY SALES SUMMARY")
    print("="*50)
    print(df.to_string(index=False))
    print("="*50)

# Main program
def main():
    # Initialize collection to store prices
    item_prices = []  # Python list collection to store prices
    
    print("🏪 COMMUNITY SHOP SALES MANAGEMENT SYSTEM 🏪")
    print("-" * 50)
    
    # Loop to collect multiple item prices
    while True:
        try:
            user_input = input("\nEnter item price (or 'done' to finish): $")
            
            # Check if user wants to finish
            if user_input.lower() == 'done':
                break
            
            # Convert input to float
            price = float(user_input)
            
            # Conditional statement to check if price is positive
            if price > 0:
                item_prices.append(price)  # Add to collection
                print(f"✅ Price ${price:.2f} recorded successfully!")
            else:
                print("❌ Error: Please enter a positive price.")
                
        except ValueError:
            print("❌ Error: Please enter a valid number or 'done' to finish.")
    
    # Check if any prices were entered
    if item_prices:
        # Call user-defined function with parameters
        sales_summary = calculate_sales_summary(item_prices)
        
        # Display results in table format
        display_sales_table(sales_summary)
        
        # Save results to dictionary and display
        print(f"\n📊 Sales Data Dictionary:")
        print(f"Raw sales summary: {sales_summary}")
        
        # Additional pandas analysis
        print(f"\n📈 DETAILED PRICE ANALYSIS:")
        prices_df = pd.DataFrame(item_prices, columns=['Item Prices'])
        print(prices_df.describe())
        
        # Save to CSV file (optional)
        try:
            prices_df.to_csv('daily_sales.csv', index=False)
            print(f"\n💾 Sales data saved to 'daily_sales.csv'")
        except:
            print(f"\n⚠️  Could not save to file")
            
    else:
        print("\n❌ No sales recorded today.")

# Run the program
if __name__ == "__main__":
    main()
```
