````python
```python
import pandas as pd

# User-defined function
def calculate_sales(prices):
    total = sum(prices)

    if len(prices) > 0:
        average = total / len(prices)
    else:
        average = 0

    return total, average


# List to store item prices
prices = []

print("===== DAILY SALES MANAGEMENT SYSTEM =====")

# Allow the user to enter multiple prices
while True:
    price_input = input("Enter item price (or type 'done' to finish): ")

    if price_input.lower() == "done":
        break

    try:
        price = float(price_input)

        # Check if price is positive
        if price > 0:
            prices.append(price)
            print("Price recorded successfully.")
        else:
            print("Invalid price! Price must be greater than 0.")

    except ValueError:
        print("Invalid input! Please enter a valid number.")


# Calculate total and average
total, average = calculate_sales(prices)


# Store the sales summary in a dictionary
sales_summary = {
    "Number of Items": len(prices),
    "Total Sales": total,
    "Average Sales": average
}


# Convert dictionary into a Pandas DataFrame
sales_table = pd.DataFrame([sales_summary])


# Display the sales summary
print("\n===== SALES SUMMARY =====")
print(sales_table.to_string(index=False))


# Save the table into a CSV file
sales_table.to_csv("sales_summary.csv", index=False)

print("\nSales summary has been saved to 'sales_summary.csv'.")
```
````
